```python
from modules.wire import INR
import torch
from math import sqrt
```

```python
def make_model(hidden_layers:int):
    return INR(
        in_features=2,
        hidden_features=256*sqrt(2),
        hidden_layers=hidden_layers,
        out_features=3,
        scale=2.,
        first_omega_0=15.,
        hidden_omega_0=15.,
    )

models = [make_model(i) for i in range(1, 7)]
```

```python
inputs_0 = torch.rand((5, 2))
inputs_1 = -1 + 2* torch.rand((5, 2))
```

```python
outs_0 = [model(inputs_0) for model in models]
outs_1 = [model(inputs_1) for model in models]
```

```python
layers = list(models[-1].net)
```

```python
h = inputs_0
outs = [h]
for layer in layers:
    h = layer(h)
    print(h.norm(dim=-1))
    outs.append(h)
```

```python
pre_activation = outs[2]@layers[2].linear.weight.T + layers[2].linear.bias
pre_activation.real.norm(dim=-1), pre_activation.imag.norm(dim=-1)
```

```python
freq = 1j * layers[2].omega_0*pre_activation
print(freq.real.norm(dim=-1), freq.imag.norm(dim=-1))
freq = torch.exp(freq)
print(freq.real.norm(dim=-1), freq.imag.norm(dim=-1))
```

```python
gauss = layers[2].scale_0*pre_activation
print(gauss.real.norm(dim=-1), gauss.imag.norm(dim=-1))
gauss = torch.exp(-torch.square(torch.abs(gauss)))
print(gauss.norm(dim=-1))
```

```python
gauss * freq
```

```python
their_way = 1j * layers[2].omega_0*pre_activation - torch.square(torch.abs(layers[2].scale_0*pre_activation))
print(their_way.real.norm(dim=-1), their_way.imag.norm(dim=-1))
their_way = torch.exp(their_way)
print(their_way.real.norm(dim=-1), their_way.imag.norm(dim=-1))
```
